# Connecting to DynamoDB table

```python
import os
import time
import boto3
from dotenv import load_dotenv
from datetime import datetime

load_dotenv()

AWS_ACCESS_KEY_ID = os.getenv("AWS_ACCESS_KEY_ID")
AWS_SECRET_ACCESS_KEY = os.getenv("AWS_SECRET_ACCESS_KEY")
AWS_REGION_NAME = os.getenv("AWS_REGION_NAME")


def dynamodb_table_connector():
    # Initialize a session using Amazon DynamoDB
    session = boto3.Session(
        aws_access_key_id=AWS_ACCESS_KEY_ID,
        aws_secret_access_key=AWS_SECRET_ACCESS_KEY,
        region_name=AWS_REGION_NAME
    )

    # Get the service resource
    dynamodb = session.resource('dynamodb')

    # Specify the DynamoDB table
    table = dynamodb.Table('whatsapp-chatbot')

    return table
```

# Adding an attribute to DynamoDB

```python
table = dynamodb_table_connector()
current_time = int(time.time())

# Convert Unix timestamp to datetime object
dt = datetime.fromtimestamp(current_time)

# Format datetime object to a human-readable string
date_time = dt.strftime('%Y-%m-%d %H:%M:%S')

wsp_id = '56942796937'
thread_id = 'thread_jb244FcgA0shNyHaR6hwzyGK'


# Define the data to be inserted
item = {
    'chatbot_partition_key': wsp_id,
    'chatbot_sort_key': 'chatbot-key',
    'thread_id': thread_id,
    'created_at': date_time,
    'time_to_live': current_time + 86400
}

# Insert the data into the table
table.put_item(Item=item)

print("Data has been saved to DynamoDB and will be automatically deleted after 1 day.")
```

# Listing every item in the table

```python
# Scan the table to get all items
response = table.scan()

# Print all items
items = response['Items']
for item in items:
    print(item)
```

# Listing an specific item in the table

```python
# Define the key of the item you want to search for
chatbot_partition_key = '56942796937'

# Query the table for the specific item
response = table.query(
    KeyConditionExpression=boto3.dynamodb.conditions.Key('chatbot_partition_key').eq(chatbot_partition_key)
)

# Print the found items
items = response['Items']
for item in items:
    print(item)
```
